```python
df = spark.table("hospital_poc.gold.patient_clinical_summaries")

display(df.limit(5))
```

```python
print("total gold records: ",df.count())
```

```python
output_path = "/Volumes/hospital_poc/gold/knowledge_documents"

rows = df.collect()

for row in rows:
    content = f"""# Patient Clinical Summary

## Patient ID
{row.patient_id}

## Specialty
{row.specialty}

## Diagnoses
{row.diagnoses}

## Medications
{row.medications}

## Investigations
{row.investigations}

## Vitals and Examination
{row.vitals_and_exam}

## Hospital Course
{row.hospital_course}

## Discharge Condition
{row.discharge_condition}

## Next Steps
{row.next_steps}

## Clinical Summary
{row.clinical_summary}
"""

    file_path = f"{output_path}/patient_{row.patient_id}.md"

    dbutils.fs.put(
        file_path,
        content,
        overwrite=True
    )

print(f"Exported {len(rows)} patient documents.")
```
