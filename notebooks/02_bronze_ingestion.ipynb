```python
from pyspark.sql import functions as F

source_path = "/Volumes/hospital_poc/bronze/raw_data/synthetic_notes_medium_1000.json"

bronze_table = "hospital_poc.bronze.patient_notes"
```

```python
df_raw = (
    spark.read
    .option("multiLine", "true")
    .json(source_path)
)

display(df_raw.limit(10))
```

```python
df_raw.printSchema()
```

```python
df_bronze = (
    df_raw
    .withColumn("ingested_at", F.current_timestamp())
    .withColumn("source_file", F.lit("synthetic_notes_medium_1000.json"))
)
```

```python
(
    df_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(bronze_table)
)
```

```python
display(
    spark.table(bronze_table).limit(10)
)
```

```python
spark.table(bronze_table).count()
```

```python
%sql
SELECT COUNT(*) AS total_records
FROM hospital_poc.bronze.patient_notes;
```

```python
%sql
--check duplicates
SELECT 
    id,
    COUNT(*) AS cnt
FROM hospital_poc.bronze.patient_notes
GROUP BY id
HAVING COUNT(*) > 1;
```

```python
%sql
--specialty distribution
SELECT
    specialty,
    COUNT(*) AS record_count
FROM hospital_poc.bronze.patient_notes
GROUP BY specialty
ORDER BY record_count DESC;
```
